````python
import pandas as pd

df= pd.read_csv("Amazon Sale Report.csv")
print(df.columns.tolist())
````

````python
# Remove unwanted columns
df = df.loc[:, ~df.columns.str.startswith("Unnamed:")]

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Check the dataset
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

df.head()
````

````python
df.info()

print("\nMissing values in each column:")
print(df.isnull().sum())
````

````python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check missing values
print("\nMissing values:")
print(df.isnull().sum().sort_values(ascending=False))

# Check order statuses
print("\nOrder statuses:")
print(df["Status"].value_counts())

# Check date range
df["Date"] = pd.to_datetime(
    df["Date"], errors="coerce"
)

print("\nDate range:")
print("Start:", df["Date"].min())
print("End:", df["Date"].max())

# Check the extra index column
print("\nFirst 10 index values:")
print(df["index"].head(10).tolist())
````

````python
# Convert dates using the known format
df["Date"] = pd.to_datetime(
    df["Date"],
    format="%m-%d-%y",
    errors="coerce"
)

# Remove the redundant row-index column
df = df.drop(columns=["index"], errors="ignore")

# Remove duplicate rows, if any
df = df.drop_duplicates()

# Check the results
print("Dataset shape:", df.shape)
print("Duplicate rows:", df.duplicated().sum())
print("Missing dates:", df["Date"].isna().sum())

print("\nDate range:")
print(df["Date"].min(), "to", df["Date"].max())
````

````python
# Check data types and missing values in key columns
key_columns = [
    "Order ID",
    "Date",
    "Status",
    "Category",
    "Qty",
    "currency",
    "Amount",
    "ship-state"
]

print(df[key_columns].isnull().sum())

print("\nQuantity data type:", df["Qty"].dtype)
print("Amount data type:", df["Amount"].dtype)

print("\nNegative quantities:", (df["Qty"] < 0).sum())
print("Zero quantities:", (df["Qty"] == 0).sum())
print("Negative amounts:", (df["Amount"] < 0).sum())
````

````python
#Investigate missing amounts and zero quantities
# Check quantity against order status
print("Quantity by order status:")
print(
    df.groupby("Status")["Qty"]
      .agg(["count", "sum"])
      .sort_values("count", ascending=False)
)

print("\nZero-quantity rows by status:")
print(
    df[df["Qty"] == 0]["Status"]
      .value_counts()
)

print("\nMissing amounts by status:")
print(
    df[df["Amount"].isna()]["Status"]
      .value_counts()
)

print("\nCurrency values:")
print(df["currency"].value_counts(dropna=False))

print("\nSample rows with missing amounts:")
display(
    df[df["Amount"].isna()][
        ["Order ID", "Date", "Status", "Category", "Qty", "currency", "Amount"]
    ].head(10)
)
````

````python
#Prepare the cleaned dataset for SQL
# Create a separate copy for the SQL database
sql_df = df.copy()

# Standardize column names for SQL and Python
sql_df.columns = (
    sql_df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)

# Rename amount to make its meaning explicit
sql_df = sql_df.rename(columns={"amount": "order_value"})

# Represent missing location as Unknown
sql_df["ship_state"] = sql_df["ship_state"].fillna("Unknown")

# Verify the cleaned structure
print("Dataset shape:", sql_df.shape)
print("\nColumns:")
print(sql_df.columns.tolist())

print("\nMissing values in important fields:")
print(sql_df[["order_id", "date", "status", "qty",
              "currency", "order_value", "ship_state"]].isnull().sum())

sql_df.head()
````

````python
#Creating SQLite database
import sqlite3

# Create a SQLite database file
conn = sqlite3.connect("amazon_sales.db")

# Save the cleaned dataset as a SQL table
sql_df.to_sql(
    "amazon_sales",
    conn,
    if_exists="replace",
    index=False
)

print("SQLite database created successfully!")

# Verify the table
cursor = conn.cursor()

cursor.execute("SELECT COUNT(*) FROM amazon_sales")
print("Rows in SQL table:", cursor.fetchone()[0])

cursor.execute("PRAGMA table_info(amazon_sales)")
print("\nSQL table columns:")
for column in cursor.fetchall():
    print(column[1], "-", column[2])
````

````python
query = """
SELECT
    category,
    COUNT(*) AS total_rows,
    SUM(qty) AS total_quantity,
    ROUND(SUM(order_value), 2) AS total_order_value
FROM amazon_sales
WHERE status != 'Cancelled'
  AND order_value IS NOT NULL
GROUP BY category
ORDER BY total_order_value DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

display(result)
````

````python
#validate and execute the generated SQL
import re
import pandas as pd

def run_safe_sql(sql):
    # Remove optional Markdown fences
    sql = sql.strip()
    sql = re.sub(r"^```(?:sql)?\s*|\s*```$", "", sql,
                 flags=re.IGNORECASE).strip()

    # Allow one SELECT query only
    if not re.match(r"^SELECT\b", sql, re.IGNORECASE):
        raise ValueError("Only SELECT queries are allowed.")

    # Reject multiple statements
    if ";" in sql.rstrip(";"):
        raise ValueError("Only one SQL statement is allowed.")

    # Block comments and common unsafe keywords
    if re.search(r"--|/\*|\*/", sql):
        raise ValueError("SQL comments are not allowed.")

    if re.search(
        r"\b(INSERT|UPDATE|DELETE|DROP|ALTER|CREATE|REPLACE|"
        r"ATTACH|DETACH|PRAGMA|VACUUM|INTO|UNION)\b",
        sql,
        re.IGNORECASE
    ):
        raise ValueError("Unsafe SQL pattern detected.")

    # Require the approved table
    if not re.search(r"\bFROM\s+amazon_sales\b", sql, re.IGNORECASE):
        raise ValueError("Query must use amazon_sales.")

    # Enforce a maximum result size
    if not re.search(r"\bLIMIT\s+\d+\b", sql, re.IGNORECASE):
        sql = sql.rstrip(";") + " LIMIT 100"

    # Open SQLite in read-only mode
    from pathlib import Path
    from urllib.parse import quote

    db_uri = (
        "file:" + quote(str(db_path.resolve()).replace("\\", "/"),
                        safe="/:") + "?mode=ro"
    )

    with sqlite3.connect(db_uri, uri=True) as read_conn:
        read_conn.execute("PRAGMA query_only = ON")
        return pd.read_sql_query(sql, read_conn)
````

````python
#Execute the generated query
# Use the SQL generated in Step 14
generated_sql = response.content

# Execute only after validation
result = run_safe_sql(generated_sql)

display(result)
````

````python
#Make the assistant answer different questions
def ask_sql_assistant(question):
    # Generate SQL from the user's question
    prompt_value = sql_prompt.invoke({
        "schema": db.get_table_info(),
        "question": question
    })

    response = llm.invoke(prompt_value)
    generated_sql = response.content.strip()

    print("Generated SQL:")
    print(generated_sql)

    # Validate and execute safely
    result = run_safe_sql(generated_sql)

    print("\nQuery results:")
    display(result)

    return result
````

````python

#Correct cancellation rate by fulfilment

corrected_sql = """
SELECT
    fulfilment,
    COUNT(*) AS total_records,
    SUM(CASE WHEN status = 'Cancelled' THEN 1 ELSE 0 END)
        AS cancelled_records,
    100.0 * SUM(
        CASE WHEN status = 'Cancelled' THEN 1 ELSE 0 END
    ) / COUNT(*) AS cancellation_percentage
FROM amazon_sales
GROUP BY fulfilment
ORDER BY cancellation_percentage DESC;
"""

result = run_safe_sql(corrected_sql)
display(result)
````
